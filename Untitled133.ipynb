```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt
import os

try:
    from torch.func import functional_call
except ImportError:
    from torch.nn.utils.stateless import functional_call

os.makedirs("loss_landscapes", exist_ok=True)
os.makedirs("spline_shapes", exist_ok=True)

# ==========================================
# 1. B-Spline Логика
# ==========================================
KNOTS = torch.linspace(0, 10, 6)

def eval_bspline_loss(x, C):
    loss = torch.zeros_like(x)
    for i in range(len(KNOTS)-1):
        if i < len(KNOTS)-2:
            mask = (x >= KNOTS[i]) & (x < KNOTS[i+1])
        else:
            mask = (x >= KNOTS[i])
        t = (x - KNOTS[i]) / (KNOTS[i+1] - KNOTS[i] + 1e-8)
        t = torch.clamp(t, 0.0, 1.0)
        seg_loss = (1 - t) * C[:, i] + t * C[:, i+1]
        loss = loss + mask.float() * seg_loss
    return loss

# ==========================================
# 2. Архитектура моделей
# ==========================================
class BaseDNN(nn.Module):
    def __init__(self, input_dim=4):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, 32), nn.ReLU(),
            nn.Linear(32, 32), nn.ReLU(),
            nn.Linear(32, 1)
        )
    def forward(self, x): return self.net(x)

class SplineMetaModel(nn.Module):
    def __init__(self, hist_len=20):
        super().__init__()
        self.conv1 = nn.Conv1d(1, 16, 3, padding=1)
        self.conv2 = nn.Conv1d(16, 16, 3, padding=1)
        self.fc1 = nn.Linear(16, 32)
        self.fc2 = nn.Linear(32, 4)
        nn.init.zeros_(self.fc2.weight)
        nn.init.zeros_(self.fc2.bias)

    def forward(self, history):
        x = history.unsqueeze(1)
        x = torch.relu(self.conv1(x))
        x = torch.relu(self.conv2(x))
        x = x.mean(dim=2)
        x = torch.relu(self.fc1(x))
        deltas = torch.sigmoid(self.fc2(x)) * 4.0 - 2.0
        base = torch.tensor([4.0, 6.0, 8.0, 10.0], device=x.device)
        C_adj = base + deltas * base
        C_fixed = torch.tensor([0.0, 2.0], device=x.device).unsqueeze(0)
        C = torch.cat([C_fixed, C_adj], dim=1)
        return C

# ==========================================
# 3. Визуализация
# ==========================================
def plot_spline_shape(C, epoch):
    C_np = C.detach().cpu().numpy()[0]
    x_grid = np.linspace(0, 12, 100)
    x_t = torch.tensor(x_grid, dtype=torch.float32)
    with torch.no_grad():
        y_t = eval_bspline_loss(x_t, C)
    plt.figure(figsize=(4, 4))
    plt.plot(x_grid, y_t.numpy(), 'b-', linewidth=2, label='B-Spline Loss')
    plt.plot(x_grid, x_grid, 'r--', alpha=0.5, label='Standard MSE')
    plt.scatter(KNOTS.numpy(), C_np, c='red', s=50, zorder=5, label='Control Points')
    plt.title(f"Spline Shape (Epoch {epoch})")
    plt.xlabel("Per-Sample MSE"); plt.ylabel("Spline Loss")
    plt.xlim(0, 12); plt.ylim(-5, 15); plt.legend(); plt.grid(True, alpha=0.3)
    plt.savefig(f"spline_shapes/spline_{epoch}.png"); plt.close()

def plot_landscape(model, X, y, title, filename):
    model.eval()
    dir1 = [torch.randn_like(p) for p in model.parameters()]
    dir2 = [torch.randn_like(p) for p in model.parameters()]
    norm1 = torch.sqrt(sum(torch.sum(p**2) for p in dir1)) + 1e-8
    norm2 = torch.sqrt(sum(torch.sum(p**2) for p in dir2)) + 1e-8
    dir1 = [p / norm1 for p in dir1]; dir2 = [p / norm2 for p in dir2]
    original_params = [p.clone() for p in model.parameters()]
    grid_range = np.linspace(-2, 2, 50)
    loss_grid = np.zeros((50, 50))
    with torch.no_grad():
        for i, a in enumerate(grid_range):
            for j, b in enumerate(grid_range):
                for idx, p in enumerate(model.parameters()):
                    p.data = original_params[idx] + a * dir1[idx] + b * dir2[idx]
                preds = model(X)
                loss_grid[i, j] = torch.mean((preds - y) ** 2).item()
    for idx, p in enumerate(model.parameters()):
        p.data = original_params[idx]
    model.train()
    plt.figure(figsize=(6, 5))
    plt.contourf(grid_range, grid_range, np.log10(loss_grid + 1e-8), levels=20, cmap='viridis')
    plt.colorbar(label='Log10(MSE)')
    plt.title(title); plt.xlabel("Direction 1"); plt.ylabel("Direction 2")
    plt.savefig(filename); plt.close()

# ==========================================
# 4. Данные
# ==========================================
np.random.seed(42)
torch.manual_seed(42)

def generate_data(n):
    X = np.random.uniform(-2, 2, (n, 4))
    y = np.sin(X[:, 0:1]) + X[:, 1:2]**2 - 0.5 * X[:, 2:3]
    return X, y

X_train_clean, y_train_clean = generate_data(300)
X_outliers = np.random.uniform(-2, 2, (75, 4))
y_outliers = np.random.uniform(-5, 5, (75, 1)) + 20.0
X_train = np.vstack([X_train_clean, X_outliers])
y_train = np.vstack([y_train_clean, y_outliers])
idx = np.random.permutation(len(X_train))
X_train, y_train = X_train[idx], y_train[idx]

X_val, y_val = generate_data(100)
X_test, y_test = generate_data(100)

X_train_t = torch.tensor(X_train, dtype=torch.float32)
y_train_t = torch.tensor(y_train, dtype=torch.float32)
X_val_t = torch.tensor(X_val, dtype=torch.float32)
y_val_t = torch.tensor(y_val, dtype=torch.float32)
X_test_t = torch.tensor(X_test, dtype=torch.float32)
y_test_t = torch.tensor(y_test, dtype=torch.float32)

# ==========================================
# 5. Эксперимент (С УСИЛЕННЫМ ADAM)
# ==========================================
EPOCHS = 1000
WARMUP_EPOCHS = 50
PLOT_EVERY = 100
LR = 1e-1

model_baseline = BaseDNN(input_dim=4)
# --- УСИЛЕНИЕ ADAM ---
# 1. Используем AdamW для правильного Weight Decay
# 2. Ставим weight_decay=1e-3 для регуляризации
optimizer_base = optim.AdamW(model_baseline.parameters(), lr=1e-2, weight_decay=1e-3)

model_spline = BaseDNN(input_dim=4)
model_meta = SplineMetaModel()
optimizer_spline = optim.SGD(model_spline.parameters(), lr=LR)
optimizer_meta = optim.Adam(model_meta.parameters(), lr=1e-3)

param_names = [n for n, _ in model_spline.named_parameters()]
params_list = list(model_spline.parameters())
history_loss = torch.zeros(20)

print("Starting training: Hardened AdamW  vs  SGD+B-Spline Loss...")
for epoch in range(EPOCHS):

    # --- 1. BASELINE (УСИЛЕННЫЙ) ---
    optimizer_base.zero_grad()
    loss_base = torch.mean((model_baseline(X_train_t) - y_train_t) ** 2)
    loss_base.backward()

    # --- УСИЛЕНИЕ ADAM: ЖЕСТКИЙ GRADIENT CLIPPING ---
    # Обрезаем норму градиента до 1.0, чтобы выбросы не уносили веса
    torch.nn.utils.clip_grad_norm_(model_baseline.parameters(), max_norm=1.0)

    optimizer_base.step()

    # --- 2. SPLINE LOSS ---
    optimizer_meta.zero_grad()
    with torch.no_grad():
        hist_input = history_loss.unsqueeze(0)
    C = model_meta(hist_input)
    current_C = C

    preds1 = model_spline(X_train_t)
    mse_per_sample = (preds1 - y_train_t) ** 2
    mse_clamped = torch.clamp(mse_per_sample, 0.0, 10.0)

    if epoch < WARMUP_EPOCHS:
        shaped_loss = torch.mean(eval_bspline_loss(mse_clamped, C))
        g_shaped = torch.autograd.grad(shaped_loss, params_list)
        with torch.no_grad():
            for p, g in zip(params_list, g_shaped): p.sub_(LR * g)
        history_loss = torch.cat([history_loss[1:], torch.mean(mse_per_sample).detach().reshape(1)])
    else:
        shaped_loss = torch.mean(eval_bspline_loss(mse_clamped, C))
        g_shaped = torch.autograd.grad(shaped_loss, params_list, create_graph=True)

        new_params = {n: p - LR * g for n, p, g in zip(param_names, params_list, g_shaped)}
        preds2 = functional_call(model_spline, new_params, X_val_t)
        meta_loss = torch.mean((preds2 - y_val_t) ** 2)
        meta_loss.backward()
        optimizer_meta.step()

        with torch.no_grad():
            C_new = model_meta(hist_input)

        preds_act = model_spline(X_train_t)
        mse_act = (preds_act - y_train_t) ** 2
        mse_act_clamped = torch.clamp(mse_act, 0.0, 10.0)

        shaped_act = torch.mean(eval_bspline_loss(mse_act_clamped, C_new))
        g_act = torch.autograd.grad(shaped_act, params_list)
        with torch.no_grad():
            for p, g in zip(params_list, g_act): p.sub_(LR * g)

        history_loss = torch.cat([history_loss[1:], torch.mean(mse_act).detach().reshape(1)])
        current_C = C_new

    if epoch % PLOT_EVERY == 0 or epoch == EPOCHS - 1:
        with torch.no_grad():
            test_base = torch.mean((model_baseline(X_test_t) - y_test_t) ** 2).item()
            test_spline = torch.mean((model_spline(X_test_t) - y_test_t) ** 2).item()

        print(f"Ep {epoch:4d} | AdamW Test: {test_base:.4f} | Spline Test: {test_spline:.4f} | C: {current_C.detach().numpy().round(2)}")

        plot_landscape(model_baseline, X_test_t, y_test_t, f"Hardened AdamW (Ep {epoch})\nTest: {test_base:.4f}", f"loss_landscapes/baseline_{epoch}.png")
        plot_landscape(model_spline, X_test_t, y_test_t, f"SGD + B-Spline (Ep {epoch})\nTest: {test_spline:.4f}", f"loss_landscapes/spline_{epoch}.png")
        plot_spline_shape(current_C, epoch)

print("\n=== ФИНАЛЬНЫЙ РЕЗУЛЬТАТ (ЧЕСТНОЕ СРАВНЕНИЕ) ===")
print(f"Hardened AdamW Test MSE:      {test_base:.4f}")
print(f"SGD + B-Spline Test MSE:      {test_spline:.4f}")
```
